```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
x= np.array([1.00000000e-17, 2.78255940e-17, 7.74263683e-17, 2.15443469e-16,
   5.99484250e-16, 1.66810054e-15, 4.64158883e-15, 1.29154967e-14,
   3.59381366e-14, 1.00000000e-13]) * 1000000000000000.0

y=np.array([2.76587972e+29, 2.76580486e+29 ,2.76522538e+29, 2.76075051e+29,
   2.72677457e+29, 2.49548832e+29, 1.62148809e+29, 7.05065718e+28,
   2.69282827e+28,9.88389557e+27])  * 1e-30
# def softplus(x, a, b, c):
    # return np.log(a + (1 + (10**x - b)**(1 / c))**(-c))
def softplus(x, a, b, c):
    return a * (1 + (b * x)**(1 / c))**(-c)
p0 = [0.2, 0.4, 0.5]
plt.loglog(x, y)
plt.loglog(x, softplus(x, *p0), label='softplus', color='red')
```

```python
popt, pcov = curve_fit(softplus, x, y, p0)
print(popt)
plt.loglog(x, softplus(x, *p0), label='softplus', color='red')
plt.loglog(x, softplus(x, *popt), label='softplus', color='green')
plt.loglog(x, y)
```
